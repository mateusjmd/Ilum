# Introdução

O presente *Jupyter Notebook* realiza um experimento estatístico com o objetivo de ilustrar e aplicar conceitos fundamentais de estatística descritiva, amostragem e inferência sob distribuições assimétricas. O exercício consiste em, a partir de uma hipótese específica, construir um experimento computacional que busque validá-la e analisar a distribuição dos estimadores obtidos por meio de simulação de Monte Carlo.
um um conjunto de dados numéricos representando uma variável aleatória, tratá-lo como a população de interesse e, a partir dele, realizar múltiplas amostragens aleatórias sem reposição para analisar o comportamento das médias amostrais em comparação à média populacional.

**Hipótese**

A posição final das partículas submetidas ao Movimento Browniano é normalmente distribuída em cada coordenada, e a distância ao ponto inicial segue distribuições relacionadas ao qui-quadrado.

Por fim, seguem as definições dos conceitos estatísticos utilizados:
> **Variável Aleatória**
>
> Associação entre cada possível resultado de um experimento aleatório e um valor numérico, cuja distribuição depende das características do fenômeno estudado.

> **Média Aritmética $(\mu)$**  
> 
> Medida de tendência central obtida pela soma dos valores observados dividida pelo número total de observações.
> $$
\mu = \frac{1}{N}\sum_{i=1}^N X_i
> $$

> **Variância $(\sigma^2)$**
> 
> Medida de dispersão que expressa o grau de afastamento dos dados em relação à média.
> $$
\sigma^2 = \mathrm{Var}(X) = \frac{1}{N} \sum_{i=1}^{N} (X_i - \mu)^2
> $$

> **Distribuição Normal**
>
> Também denominada de 'Distribuição Gaussiana' é a distribuição de uma variável aleatória $X$ com média $\mu \in \mathbb{R}$ e variância $\sigma^2 > 0$, denotada por $X \sim \mathcal{N}(\mu, \sigma^2)$ se sua função de densidade de probabilidade é dada por:
> $$
f_X(x) = \frac{1}{\sigma\sqrt{2\pi}} \exp{\left( - \frac{(x - \mu)^2}{2\sigma^2} \right)}, \; x \in \mathbb{R}
> $$

> **Distribuição de Rayleigh**
>
> É a distribuição relativa ao módulo de um vetor bidimensional cujas componentes são variáveis aleatórias gaussianas independentes e de mesma variância, tendo a função de densidade de probabilidade dada por:
> $$
f(r; \sigma) = \begin{cases}
\frac{r}{\sigma^2}e^{-\frac{r^2}{2\sigma^2}}, r \geq 0 \\
0, r < 0
\end{cases}
> $$

> **Distribuição de Maxwell-Boltzmann**
>
> É a distribuição relativa ao módulo da velocidade (ou deslocamento) de uma partícula quando cada componente segue uma distribuição normal independente, cuja função de densidade de probabilidade é dada por:
> $$
f(v; \sigma) = \begin{cases}
\sqrt{\frac{2}{\pi}} \frac{v^2}{\sigma^3} e^{-\frac{v^2}{2\sigma^2}}, v \geq 0 \\
0, r < 0
\end{cases}
> $$

> **Vetor Gaussiano**
>
> Vetor cujas componentes são variáveis gaussianas. Assim, se $Z = (Z_1, Z_2, ..., Z_d)$, e cada $Z_i \sim \mathcal{N}(0, 1)$ independentes , então:
> $$
Z \sim \mathcal{N}(0, I_d)
>$$
> - 0: Vetor nulo (Média em cada dimensão é 0)
> - $I_d$: Matriz identidade $d \times d$, que indica covariância unitária e independência entre as componentes
>
> Nas simulações, esse vetor define direções aleatórias em $d$ dimensões.

> **Esfera $S^{d-1}$**
>
> Esfera unitária em $d$ dimensões, dada por:
> $$
S^{d-1} = \{x \in \mathbb{R}^d: ||x|| = 1\}
> $$
> A dimensão da esfera é sempre uma unidade menor que a dimensão do espaço:
> - Círculo $S^1$ está em $\mathbb{R}^2$
> - Esfera $S^2$ está em $\mathbb{R}^3$
> - Hiperesfera $S^{d-1}$ está em $\mathbb{R}^d$
> 
> Na simulação, gerar uma direção aleatória significa escolher um ponto uniforme em $S^{d-1}$. Isso é feito pela normalização de um vetor gaussiano $Z \sim \mathcal{N}(0, I_d)$:
> $$
U = \frac{Z}{||Z||}
>$$ 

> **Valor Esperado**
>
> Para qualquer função $g(X)$, o valor esperado é:
>
> $$
\mathbb{E}[g(X)] = \begin{cases}
\sum_i g(x_i)p_i, \;\;\; \text{se X for discreta}, \\
\int_{-\infty}^{+\infty} g(x)f(x) dx,\;\;\; \text{se X for contínua}
\end{cases}
> $$

> **Deslocamento Quadrático Médio (MSD - *Mean Squared Displacement*)**
>
> É a medida do crescimento médio da distância ao longo do tempo, usado para diferenciar regimes:
> - Linear (difusivo): $MSD \sim t$
> - Quadrático (balístico): $MSD \sim t^2$
> - Saturado: Confinamento
>
> O MSD é dado por:
> $$
MSD(n) = \frac{1}{M} \sum_{j=1}^M ||X_n^{(j)}||^2
> $$
> - M: Número de simulações independentes

> **Equação Diferencial Estocástica (SDE - *Stochastic Differential Equation*)**
>
> Equações diferenciais estocásticas são aquelas dadas, de modo geral, pela forma:
> $$
dX(t, \omega) = f(t, X(t, \omega))dt + g(t, X(t, \omega))dW(t, \omega)
> $$


> **Integração de Euler**
>
> Método de integração numérica para SDE's, dado por:
> $$
X_{n+1} = X_n + f(X_n, t_n) \Delta t + g(X_m, t_n) \Delta W_n, \; com \;\;\; \Delta W_n \sim \sqrt{\Delta t} \cdot \mathcal{N}(0, 1)
> $$
>
> Essa é, mais especificamente, a versão do Método de Euler adaptada a ruídos aleatórios, fundamental para simular o Movimento Browniano.

> **Entropia**
>
> Na Teoria da Informação, entropia é a medida da quantidade média de informação ou incerteza de uma variável aleatória discreta $X$ com distribuição de probabilidades $p(x)$ dada por:
> $$
H(X) = - \sum_{x \in \chi} p(x) \log (p(x))
> $$
>
> - $H(X)$: Entropia de Shannon
> - $\log (p(x))$: Pode ser em base 2 (*bits*) ou $e$ (*nats*)

> **Lei dos Grandes Números (LGN)**
>
> A soma de muitas variáveis aleatórias independentes (ou não-correlacionadas, etc.) tende a estar próxima da sua esperança. Mais precisamente, a média observada $\frac{X_1 + \cdots + X_n}{n}$, que é aleatória, se aproxima da média teórica $\mathbb{E}[\frac{X_1 + \cdots + X_n}{n}]$, que é determinística.

> **Teorema Central do Limite (TCL)**
>
> Seja $(X_n)_{n \in \mathbb{N}}$ uma sequência de variáveis aleatórias independentes e identicamente distribuídas não-degeneradas com segundo momento finito. Tome $S_n = X_1 + X_2 + \cdots + X_n$. Então:
> $$
\frac{S_n - \mathbb{E}S_n}{\sqrt{\mathbb{V}S_n}} \xrightarrow{d} \mathcal{N}(0,1)
> $$

# Metodologia

### Importação das Bibliotecas

````python
import numpy as np
import plotly.graph_objects as go
from plotly.subplots import make_subplots
from scipy.stats import norm, rayleigh, maxwell
from tabulate import tabulate
````

### Simulações

A seguinte função constrói um passeio aleatório (*random walk*) isotrópico, como um protótipo do movimento browniano em tempo discreto.

````python
def gerar_trajetoria(num_passos, dimensao=2, passo=1.0):
    """
    Gera uma trajetória browniana simplificada (passeio aleatório) em 1D, 2D ou 3D.

    Cada passo da trajetória é amostrado de forma aleatória, com direção uniforme na esfera unitária da dimensão especificada e comprimento fixo.

    ### Parâmetros
        num_pasos: Número de passos discretos na trajetória.

        dimensao: Número de dimensões do espaço no qual a trajetória ocorre.

        passo: Comprimento de cada passo.
    
    ### Retorna
        posicoes: Array de formato `(num_passos, dimensao)` contendo as posições cumulativas da trajetória em cada passo.

    """

    # Amostragem de vetores gaussianos padrão Z_k ~ N(0, I_d)
    direcoes = np.random.normal(size=(num_passos, dimensao))

    # Normaliza cada vetor para norma 1, produzindo direções uniformes na esfera
    direcoes /= np.linalg.norm(direcoes, axis=1)[:, None]

    # Cada passo tem comprimento fixo: l = passo e direção uniforme
    deslocamentos = passo * direcoes
    
    # Somua cumulativa dos incrementos -> Posição após cada passo
    posicoes = np.cumsum(deslocamentos, axis=0)

    # Retorna a trajetória
    return posicoes
````

Explicação detalhada:

Para fazer o *random walk* isotrópico, primeiro foi realizada a amostragem de vetores gaussianos padrão $Z_k \sim \mathcal{N}(0, I_d)$, pois vetores gaussianos isotrópicos têm direção distribuída uniformemente na esfera $S^{d-1}$ quando normalizados, garantindo assim a uniformidade da direção. Com isso, tem-se:
```python
    direcoes = np.random.normal(size=(num_passos, dimensao))
```

Por conseguinte, se $Z \sim \mathcal{N}(0, I_d)$, então $U = \frac{Z}{||Z||}$ é uniforme em $S^{d-1}$, de modo que isso garante a isotropia - uma vez que não há direção preferencial. Consequentemente, cada incremento terá módulo fixo definido, resultando em um *random walk* de passo fixo, o qual converge ao browniano no limite de muitos passos (Teorema do Limite Central - TCL), mas não é exatamente o browniano de incrementos gaussianos em tempo finito. Disso, tem-se:
```python
    direcoes /= np.linalg.norm(direcoes, axis=1)[:, None]
```

Então, define-se que cada passo tem um comprimento fixo ($l = passo$) e direção uniforme. Esse fato conecta-se com a difusão porque, se cada passo ocorre a cada $\Delta t$, então $\mathbb{E}||X_n||^2 = nl^2 \Rightarrow \underbrace{\mathbb{E}||X_t||^2}_{MSD} = 2dDt$ . Identificando $D$ como $D = \frac{l^2}{2s\Delta t}$, nota-se o '$D$ efetivo' do passo fixo. Assim, tem-se:
```python
    deslocamentos = passo * direcoes
```

Por fim, realiza-se a soma cumulativa de todos os incrementos, obtendo a posição após cada passo. Nesse contexto, $X_{k+1} = X_k + \Delta X_k$ é a integração de Euler (no caso browniano puro, Euler-Maruyama), mas com incrementos i.i.d., constrói-se um *random walk*. Desse modo, tem-se:
```python
    posicoes = np.cumsum(deslocamentos, axis=0)
```

Enfim, a trajetória $k \mapsto X_k$ em $\mathbb{R}^d$ é retornada:
```python
    return posicoes
```

Função para simular o experimento anteriormente delineado, rodando $N$ réplicas independentes (ensaios Monte Carlo) para estimar estatísticas de interesse. A partir disso, espera-se que a Lei dos Grandes Números (LGN) garantirá a convergência das médias amostrais.

````python
def simular_experimento(num_simulacoes, num_passos, dimensao=2, passo=1.0):
    """
    Executa múltiplas simulações de trajetórias em um espaço de dimensão arbitrária.

    Cada trajetória é gerada passo a passo por meio da função `gerar_trajetoria`, e a posição final de cada simulação é armazenada para 
    análises estatísticas posteriores.

    ### Parâmetros
        num_simulacoes: Número de simulações a serem executadas.

        num_passos: Número de passos discretos em cada trajetória.

        dimensao: Número de dimensões do espaço no qual a trajetória ocorre.

        passo: Comprimento de cada passo na simulação.
    
    ### Retorna
        Uma tupla contendo dois arrays do NumPy:
        - trajetorias : np.ndarray
            Array de formato `(num_simulacoes, num_passos, dimensao)` com todas as trajetórias simuladas.
        - posicoes_finais : np.ndarray
            Array de formato `(num_simulacoes, dimensao)` com a última posição de cada trajetória.
    """
    
    # Armazena a trajetória de cada simulação para posteriores visualizações
    trajetorias = []

    # Armazena a última posição da trajetória de cada simulação para estatísticas posteriores
    posicoes_finais = []

    for _ in range(num_simulacoes):
        traj = gerar_trajetoria(num_passos, dimensao, passo) # Executa a simulação
        trajetorias.append(traj)
        posicoes_finais.append(traj[-1])
    
    # Converte as listas anteriores em arrays de NumPy e retorna em forma de tupla para operações vetorizadas posteriores
    return np.array(trajetorias), np.array(posicoes_finais)
````

Função para cálculo de algumas métricas dos dados obtidos a partir das simulações, como média, variância e MSD:

````python
def analisar_resultados(posicoes_finais):
    """
    Calcula estatísticas básicas das posições finais de múltiplas trajetórias.

    Esta função calcula a média, variância e Mean Squared Displacement (MSD) das posições finais de um conjunto de trajetórias simuladas, 
    permitindo avaliar a dispersão e o comportamento médio do sistema.
    
    ### Parâmetros
        posicioes_finais: Array de formato `(num_simulacoes, dimensao)` contendo a última posição de cada trajetória simulada.

    ### Retorna
        Dicionário contendo as métricas de média, variância e MSD.
    """
    
    # Quadrado das posições finais
    distancias2 = np.sum(posicoes_finais**2, axis=0)
    
    # Média das posições finais
    media = np.mean(posicoes_finais, axis=0)
    
    # Variância das posições finais
    variancia = np.var(posicoes_finais, axis=0)
    
    # MSD 
    msd = np.mean(distancias2)
    
    # Retorna dicionário contendo as métricas
    return {
        'media': media,
        'variancia': variancia,
        'MSD': msd
    }
````

Função para plotar as trajetórias obtidas com as $N$ réplicas independentes da simulação:

````python
def plotar_trajetorias(dimensao, trajetorias, num_exemplos=5):
    """
    Plota trajetórias de exemplo em 1D, 2D ou 3D usando Plotly.

    A função seleciona as primeiras trajetórias fornecidas e cria um gráfico interativo correspondente à dimensão do espaço. 
    Para 3D, o ponto de origem é destacado como referência.

    ### Parâmetros
        dimensao: Dimensão do espaço das trajetórias. Aceita 1, 2 ou 3.

        trajetorias: Array de formato `(num_simulacoes, num_passos, dimensao)` contendo as trajetórias simuladas.

        num_exemplos: Número de trajetórias a serem exibidas no gráfico.


    ### Retorna
        A função exibe o gráfico interativo, mas não retorna valores.
    """
    fig = go.Figure()

    # Plot Unidimensional
    if dimensao == 1:
        for traj in trajetorias[:num_exemplos]:
            fig.add_trace(go.Scatter(y=traj[:,0], mode='lines', name='trajetória'))
        fig.update_layout(
            title=f'{num_exemplos} Trajetórias de Exemplo (1D)',
            xaxis_title='Passo',
            yaxis_title='Posição',
            width=1000,
            height=600,
            plot_bgcolor='white')

        # Configuração do background
        fig.update_xaxes(showgrid=True, gridcolor='white', zerolinecolor='darkgray')
        fig.update_yaxes(showgrid=True, gridcolor='white', zerolinecolor='darkgray')

    # Plot Bidimensional
    elif dimensao == 2:
        for traj in trajetorias[:num_exemplos]:
            fig.add_trace(go.Scatter(x=traj[:,0], y=traj[:,1], mode='lines', name='trajetória'))

        fig.update_layout(
            title=f'{num_exemplos} Trajetórias de Exemplo (2D)',
            xaxis_title='X',
            yaxis_title='Y',
            width=1000,
            height=600,
            plot_bgcolor='white')
        
        # Configuração do background
        fig.update_xaxes(showgrid=True, gridcolor='white', zerolinecolor='darkgray')
        fig.update_yaxes(showgrid=True, gridcolor='white', zerolinecolor='darkgray')        

    # Plot Tridimensional    
    elif dimensao == 3:
        for i, traj in enumerate(trajetorias[:num_exemplos]):
            fig.add_trace(go.Scatter3d(x=traj[:,0], y=traj[:,1], z=traj[:,2], mode='lines', name=f'Trajetória {i+1}'))
        
        # Ponto da origem (X=0, Y=0, Z=0)
        fig.add_trace(go.Scatter3d(
            x=[0], y=[0], z=[0],
            mode='markers',
            marker=dict(size=6, color='Black', symbol='circle'),
            name='Origem',
            showlegend=True))
        
        # Configuração do background
        fig.update_layout(
            title=f'{num_exemplos} Trajetórias de Exemplo (3D)',
            width=1000,
            height=800,
            scene=dict(
            xaxis=dict(title='x', showgrid=True, gridcolor='white', zerolinecolor='black'),
            yaxis=dict(title='y', showgrid=True, gridcolor='white', zerolinecolor='black'),
            zaxis=dict(title='z', showgrid=True, gridcolor='white', zerolinecolor='black'),
            aspectmode='data'))

        # Ajusta a área interativa do plot 3D à área da figura
        fig.update_scenes(
            xaxis_title='x', yaxis_title='y', zaxis_title='z', aspectmode='data')

    fig.show()
````

Função para plotar a distribuição das posições e distâncias finais, a fim de comparar a visão teórica para $R = ||X_t||$:

````python
def plotar_distribuicoes(dimensao, posicoes_finais):
    """
    Plota as distribuições estatísticas das posições finais de trajetórias simuladas.

    ### Parâmetros
        dimensao: Dimensão do espaço das trajetórias. Aceita valores 1, 2 ou 3.

        posicoes_finais: Array de formato `(num_simulacoes, dimensao)` contendo a última posição de cada trajetória simulada.

    ### Retorna
        A função exibe o gráfico interativo, mas não retorna valores.
    """
    coordenada = posicoes_finais[:, 0]
    distancias = np.linalg.norm(posicoes_finais, axis=1)

    fig = make_subplots(
        rows=1, cols=2,
        subplot_titles=('Distribuição das Posições Finais Unidimensionais', 'Distribuição da Distância Final'),
        horizontal_spacing=0.15
    )

    # Histograma das coordenadas
    fig.add_trace(
        go.Histogram(x=coordenada, nbinsx=200, histnorm='probability density',
                     name='Coordenada', opacity=0.6, marker_color='royalblue'),
        row=1, col=1
    )
    fig.update_xaxes(title_text="Valor da Coordenada", row=1, col=1)
    fig.update_yaxes(title_text="Densidade de Probabilidade", row=1, col=1)

    # Normal MLE (Most likelihood estimation)
    mu_fit, sigma_fit = norm.fit(coordenada)
    x_norm = np.linspace(coordenada.min(), coordenada.max(), 600)
    fig.add_trace(
        go.Scatter(x=x_norm, y=norm.pdf(x_norm, mu_fit, sigma_fit), mode='lines',
                   name=f'Normal Ajustada',
                   line=dict(width=2)),
        row=1, col=1
    )
    

    # Histograma das distâncias euclidianas
    fig.add_trace(
        go.Histogram(x=distancias, nbinsx=200, histnorm='probability density',
                     name='Distância', opacity=0.6, marker_color='darkorange'),
        row=1, col=2
    )
    fig.update_xaxes(title_text="Distância Final", row=1, col=2)
    fig.update_yaxes(title_text="Densidade de Probabilidade", row=1, col=2)

    # ----- Curvas Teóricas -----
    # Eixo X da distribuição de distâncias finais
    x_dist = np.linspace(0, distancias.max(), 600)

    # Distribuição de Rayleigh teórica ajustada aos dados
    if dimensao == 2:
        loc, scale = rayleigh.fit(distancias, floc=0)
        fig.add_trace(
            go.Scatter(x=x_dist, y=rayleigh.pdf(x_dist, loc=loc, scale=scale), mode='lines',
                       name=f'Rayleigh Ajustada'),
            row=1, col=2
        )

    # Distribuição de Maxwell-Boltzmann teórica ajustada aos dados
    elif dimensao == 3:
        loc, scale = maxwell.fit(distancias, floc=0)
        fig.add_trace(
            go.Scatter(x=x_dist, y=maxwell.pdf(x_dist, loc=loc, scale=scale), mode='lines',
                       name=f'Maxwell-Boltzmann Ajustada'),
            row=1, col=2
        )

    fig.update_layout(width=1000, height=500, showlegend=True, bargap=0.05, plot_bgcolor='white')
    fig.show()
````

Função que unifica todas as demais, executando a simulação e os plots automaticamente:

````python
def rodar_simulacao_completa(num_simulacoes=1000, num_passos=100, dimensao=2, passo=1.0, num_exemplos=5):
    """
    Executa uma simulação completa de trajetórias brownianas, calcula estatísticas e plota visualizações das trajetórias e distribuições finais.
    ### Parâmetros
        num_simulacoes: Número de trajetórias a serem simuladas.

        num_passos: Dimensão do espaço das trajetórias (1, 2 ou 3).

        dimensao: Comprimento de cada passo na simulação.

        passo: Número de trajetórias a serem exibidas nos gráficos de exemplo.
    
    ### Retorna
        Dicionário contendo as métricas calculadas pela função `analisar_resultados()`.
    """
    # Coleta de dados via Monte Carlo
    trajetorias, posicoes_finais = simular_experimento(num_simulacoes, num_passos, dimensao, passo)

    # Calcula média, variâcia e MSD dos dados
    estatisticas = analisar_resultados(posicoes_finais)

    # Visualização
    plotar_trajetorias(dimensao, trajetorias, num_exemplos)
    plotar_distribuicoes(dimensao, posicoes_finais)

    # Retorna o dicionário contendo as estatísticas
    return estatisticas
````

Roda a simulação $n$-vezes, plotando gráfico da trajetória, distribuição uni-axial e da distância euclidiana final, bem como exibe as métricas de média, variância e MSD:

````python
resultados = rodar_simulacao_completa(num_simulacoes=1_000, num_passos=1_000, dimensao=3, passo=1.0, num_exemplos=5);

# Tabelas das métricas de Média, Variância e MSD
if len(resultados['media']) == 1: # Métricas unidimensionais
        tabela_resultados = [
            ['Média', resultados['media']],
            ['Variância', resultados['variancia']],
            ['MSD', resultados['MSD']]
        ]
        print(tabulate(tabela_resultados, headers=['Métricas', 'X'], tablefmt='github', floatfmt='.2f'))

elif len(resultados['media']) == 2: # Métricas bidimensionais
        tabela_resultados = [
            ['Média', resultados['media'][0], resultados['media'][1]],
            ['Variância', resultados['variancia'][0], resultados['variancia'][1]],
            ['MSD', resultados['MSD']]
        ]
        print(tabulate(tabela_resultados, headers=['Métricas', 'X', 'Y'], tablefmt='github', floatfmt='.2f'))
else: # Métricas n-dimensionais
        tabela_resultados = [
            ['Média', resultados['media'][0], resultados['media'][1], resultados['media'][2]],
            ['Variância', resultados['variancia'][0], resultados['variancia'][1], resultados['variancia'][2]],
            ['MSD', resultados['MSD']]
        ]
        print(tabulate(tabela_resultados, headers=['Métricas', 'X', 'Y', 'Z'], tablefmt='github', floatfmt='.2f'))
````

# Conclusão

> **Comportamento em 1D (Normal)**

Em uma dimensão, cada passo do passeio aleatório segue diretamente uma distribuição gaussiana devido à forma como foi construído (gerando incrementos gaussianos independentes).

No entanto, quando olhamos a distribuição da posição final, vemos que ela nem sempre se sobrepõe perfeitamente à curva da normal teórica.

Isso pode acontecer por dois motivos principais: número de passos finito e amostragem limitada: a Lei dos Grandes Números e o Teorema Central do Limite garantem convergência para uma gaussiana, mas somente no limite de muitas partículas e muitos passos.

> **Comportamento em 2D (Rayleigh)**

Em duas dimensões, a distribuição da magnitude do deslocamento radial não segue uma gaussiana, mas sim a distribuição de Rayleigh.

Isso ocorre porque cada componente (x e y) segue uma normal independente e a distância radial $r = \sqrt{x^2 + y^2}$ é a raiz quadrada da soma de quadrados de variáveis normais independentes. Por definição, isso leva exatamente a uma distribuição de Rayleigh.

O histograma simulado se aproxima da curva teórica, mas pode haver pequenas discrepâncias devido às mesmas limitações amostrais.

> **Comportamento em 3D (Maxwell-Boltzmann)**

Em três dimensões, o raciocínio se expande: cada componente $(x, y, z)$ é gaussiano independente, e o módulo do vetor deslocamento segue a distribuição de Maxwell-Boltzmann.

Essa distribuição descreve a probabilidade de encontrar uma partícula com certo deslocamento (ou velocidade, em física estatística) no espaço tridimensional.

O histograma se ajusta bem à curva teórica, mostrando que a simulação respeita a estatística esperada.
